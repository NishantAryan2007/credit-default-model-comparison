```python
import sys
sys.path.append("..")
import pandas as pd
from sklearn.model_selection import RepeatedStratifiedKFold, cross_validate
from src.features import split_xy
from src.models import build_dummy, build_logreg

train = pd.read_csv("../data/processed/train.csv")
X, y = split_xy(train)

cv = RepeatedStratifiedKFold(n_splits = 5, n_repeats = 2, random_state = 42)
scoring = ["roc_auc", "average_precision", "neg_log_loss", "neg_brier_score"]

for name, model in [("dummy", build_dummy()), ("logreg", build_logreg())]:
    res = cross_validate(model, X, y, cv = cv, scoring = scoring)
    print(name)
    for m in scoring:
        s = res[f"test_{m}"]
        print(f"  {m:20s} {s.mean():.4f} +/- {s.std():.4f}")
```

Logistic regression got ROC-AUC 0.77, the dummy got 0.50, so it learned something real.
