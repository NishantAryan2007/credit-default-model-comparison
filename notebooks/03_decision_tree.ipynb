```python
import sys
sys.path.append("..")
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_validate
from src.features import split_xy
from src.models import build_tree

train = pd.read_csv("../data/processed/train.csv")
X, y = split_xy(train)

cv = StratifiedKFold(n_splits = 5, shuffle = True, random_state = 42)

print("depth   train AUC   validation AUC")
for depth in [1, 2, 3, 5, 8, 12, 20, None]:
    res = cross_validate(build_tree(max_depth = depth), X, y, cv = cv,
                         scoring="roc_auc", return_train_score=True)
    print(f"{str(depth):5s}   {res['train_score'].mean():.3f}       {res['test_score'].mean():.3f}")
```

Train AUC keeps rising with depth, but validation AUC peaks at depth 5 (0.756) and then falls. That is overfitting.
