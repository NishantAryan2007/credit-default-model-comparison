```python
import sys
sys.path.append("..")
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_validate
from src.features import split_xy
from src.models import build_forest

train = pd.read_csv("../data/processed/train.csv")
X, y = split_xy(train)

cv = StratifiedKFold(n_splits = 5, shuffle = True, random_state = 42)

print("depth   min_leaf   train AUC   validation AUC")
for depth, leaf in [(3, 1), (5, 1), (8, 1), (12, 1), (None, 1), (None, 20), (None, 50)]:
    res = cross_validate(build_forest(max_depth = depth, min_samples_leaf = leaf), X, y, cv = cv,
                         scoring = "roc_auc", return_train_score = True)
    print(f"{str(depth):5s}   {leaf:<8d}   {res['train_score'].mean():.3f}       {res['test_score'].mean():.3f}")
```

Validation AUC stays between 0.76 and 0.785. Forests still overfit, but averaging protects validation. Best settings limit leaf size.
